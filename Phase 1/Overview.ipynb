# Phase 1: Build the foundation

## Overview

Phase 1 is about building the foundations being Python fluency and probability intuition.

The goal is that at the end of this phase I will feel comfortable manipulating data in pandas and NumPy, and understand basic probability from first principles.


## Step 1 — Python for Finance by Yves Hilpisch

### What to do

Work through chapters 1 to 6 of Python for Finance by Yves Hilpisch. This is will be my primary technical resource for the entire phase.

The rule for every chapter is the same — read a section, then immediately open a notebook and type out every code example myself plus further notes to accompany it. Do not copy paste. Why? The act of typing the code manually forces my brain to process what each line is actually doing rather than just recognising it visually. Futher after each chapter, add a markdown cell at the bottom of your notebook written in plain English explaining what I learned, what was confusing, and what I want to revisit.

### Chapters to read

Chapter 1 — Why Python for Finance. Short introductory chapter. Gives context for why Python became the dominant language in quantitative finance. Read it in one sitting, no code to write yet.

Chapter 2 — Python infrastructure. Setting up your environment, understanding Anaconda, Jupyter, and how the tools fit together. Follow every instruction and make sure everything runs before moving on.

Chapter 3 — Introduction to Python. Core syntax, data types, loops, functions, and basic data structures. Even if you already know Python, do not skip this — Hilpisch covers Python in the context of financial applications which is a different angle from a standard CS course.

Chapter 4 — NumPy. The numerical computing library that underpins almost all quantitative Python work. Arrays, vectorised operations, mathematical functions. Work through every example carefully — this is foundational.

Chapter 5 — Data analysis with pandas. The most important chapter in the book for your purposes. DataFrames, time series, indexing, groupby, merging, plotting. Spend extra time here. If any chapter deserves a second read it is this one. Every project you build for the next 12 months will use pandas heavily.

Chapter 6 — Object oriented programming. Classes, methods, inheritance in a financial context. Read it fully but if it slows you down significantly, skim it and come back — it is the least immediately critical chapter in this block.

## Step 3 — Flash Boys by Michael Lewis

### What to do

Read Flash Boys cover to cover alongside your technical work. This is not a textbook — it is a narrative non-fiction book that reads like a thriller. Aim for one chapter per night in the evenings after you have finished your technical work for the day.

### What it covers

Flash Boys tells the story of how high-frequency trading firms came to dominate the US stock market, and the group of people who tried to build a fairer exchange in response. It covers market microstructure — how orders actually get executed, what happens in the milliseconds between you pressing buy and the trade completing, and how speed became the most valuable commodity in modern markets.

### Why read it now

You are about to spend months learning technical material — probability, Python, derivatives pricing. Flash Boys gives you the human and cultural context for what all of that technical knowledge is actually used for in the real world. It makes the abstract concrete. It also gives you things to talk about in networking conversations and interviews — being able to reference Flash Boys naturally signals to practitioners that you have thought about markets beyond the textbook.

### Reading approach

Read it in your own time, not at your desk. This is your wind-down reading. Do not take technical notes — just read it and absorb the story. If something surprises you or sparks a question, jot a single sentence in your phone to look up later.


## Step 4 — Khan Academy probability and statistics

### What to do

Work through the entire probability and statistics track on Khan Academy. Go to khanacademy.org and search for the statistics and probability course. Complete every module in sequence from start to finish.

Do not just watch the videos. Do every exercise after every lesson. The exercises are where the learning actually happens — the videos are just the introduction.

At 1 hour per day this takes approximately 3 to 4 weeks. If you can do 90 minutes some days, do it — getting this foundation solid early pays dividends throughout the rest of the roadmap.

### Modules to complete in order

Start with basic probability — sample spaces, events, and the fundamental rules. Then move through counting and combinatorics, which covers permutations and combinations that appear constantly in quant interview problems. Then study random variables and probability distributions, covering discrete distributions like binomial and Poisson, and continuous distributions like normal and exponential. Then work through the central limit theorem and sampling distributions, followed by basic hypothesis testing and confidence intervals.

### Why probability is non-negotiable

Probability is the core language of quantitative finance. Options are priced using probability distributions. Risk is measured using probabilistic models. Trading strategies are evaluated using statistical tests. Every single concept you will encounter in the rest of this roadmap — Black-Scholes, the Sharpe ratio, factor models, Monte Carlo simulation — is built directly on top of probability theory. There is no shortcut around this. Get it solid now.

### How to take notes

For each major concept create a markdown cell in a dedicated notebook called probability-notes.ipynb. Write the definition in your own words, write out the key formula, then write a one sentence explanation of why it matters in a financial context. For example, after learning about the normal distribution you would write what it is, write out the formula for the probability density function, and then write something like — asset returns are often modelled as normally distributed, which is why standard deviation is used as a measure of investment risk.
